```python
from google.colab import drive

drive.mount('/content/drive')

import os

os.listdir('/content/drive/MyDrive')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Spotify dataset
df = pd.read_csv('/content/drive/MyDrive/Colab/Most Streamed Artists on Spotify.csv')

# View dataset
print(df.head())
print(df.columns)


# ============================================================
# 1st Graph - Number of Artists by Genre
# ============================================================

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    y="Primary Genre",
    order=df["Primary Genre"].value_counts().index
)

plt.title("Number of Artists by Primary Genre")
plt.xlabel("Count")
plt.ylabel("Primary Genre")

plt.show()


# ============================================================
# 2nd Graph - Total Streams Distribution
# ============================================================

plt.figure(figsize=(6,4))

plt.hist(
    df["Total Streams (in millions)"].dropna(),
    bins=15,
    color="skyblue",
    edgecolor="black"
)

plt.title("Total Streams Distribution")
plt.xlabel("Total Streams (in millions)")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 3rd Graph - Correlation Heatmap
# ============================================================

plt.figure(figsize=(10,7))

numeric_df = df.select_dtypes(include="number")

corr_matrix = numeric_df.corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap - Spotify Dataset")

plt.show()
```
