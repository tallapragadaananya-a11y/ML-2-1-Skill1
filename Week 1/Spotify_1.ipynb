```python
import pandas as pd
#File name
CSV_PATH = "Most Streamed Artists on Spotify.csv"
#Read CSV File
df = pd.read_csv(CSV_PATH)
#1.Display firt 5 rows
print("FIRST 5 ROWS")
print(df.head())
# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)
# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)
# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)
# 5. Display basic information
print("\nDATASET INFORMATION")
df.info()
# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe(include="all"))
# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())
# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
# 9. Most streamed artist distribution (if column exists)
possible_target_cols = ["Artist", "artist", "Artists", "name"]
target_col = next((col for col in possible_target_cols if col in df.columns), None)
if target_col:
    print(f"\n{target_col.upper()} DISTRIBUTION")
    print(df[target_col].value_counts())

    print(f"\n{target_col.upper()} PERCENTAGE")
    print(df[target_col].value_counts(normalize=True) * 100)
else:
    print("\nNo artist/name column found for distribution analysis.")
```
